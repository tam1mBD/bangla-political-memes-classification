```python
import numpy as np
import pandas as pd
# import matplotlib.pyplot as plt
# import seaborn as sns
```

```python
df = pd.read_csv('submissions/submission_ensemble.csv')
dfx = pd.read_csv('Test_with_labels.csv')
```

```python
df['Label'].value_counts()
```

```python
dfx['Label'].value_counts()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
confusion_matrix(dfx['Label'], df['Label'])
```

```python
from sklearn.metrics import accuracy_score, precision_score, f1_score, classification_report

# Calculate metrics
accuracy = accuracy_score(dfx['Label'], df['Label'])
precision_macro = precision_score(dfx['Label'], df['Label'], average='macro')
f1_macro = f1_score(dfx['Label'], df['Label'], average='macro')

print(f"{'='*50}")
print(f"MODEL EVALUATION METRICS")
print(f"{'='*50}")
print(f"Accuracy:           {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision (macro):  {precision_macro:.4f}")
print(f"F1-Score (macro):   {f1_macro:.4f}")
print(f"{'='*50}\n")

# Detailed classification report
print("Detailed Classification Report:")
print(classification_report(dfx['Label'], df['Label'], digits=4))
```
